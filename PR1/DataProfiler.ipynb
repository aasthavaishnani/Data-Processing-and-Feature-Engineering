## Question 1: What is Data Analysis?

### Definition

Data Analysis is the process of collecting, cleaning, and studying data to find useful information. It helps us understand data, find patterns, and make better decisions.

### Main Steps of Data Analysis

1. **Data Collection:** It is the process of collecting data from different sources, such as CSV files, databases, APIs, and JSON files.

2. **Data Cleaning:** It means finding and correcting errors in the data. It also includes handling missing values, duplicate records, and incorrect data formats.

3. **Exploratory Data Analysis (EDA):** It is used to understand data by using graphs, charts, and basic statistical methods. It helps us find patterns, trends, and relationships in the data.

4. **Data Interpretation:** It means understanding the results of data analysis and using them to get useful information and make decisions.

### Importance of Data Analysis in Business

Data Analysis is very useful for businesses because it helps them make better decisions based on data. It helps companies understand customer needs, improve their work, reduce problems, and predict future trends. It also helps businesses understand what is working well and what needs improvement.



## Question 2: Describe How to Plan a Data Science Project

Planning a Data Science project means deciding the steps needed to solve a problem using data. Proper planning helps us complete the project in an organized way and get useful results.

### Steps of a Data Science Project

1. **Understand the Business Problem:**

   First, we need to understand the problem that we want to solve. We should clearly decide the main goal of the project and what results we want to get.

2. **Data Collection:**

   In this step, we collect the required data from different sources, such as CSV files, databases, APIs, and websites. The collected data is used for further analysis.

3. **Data Understanding and Exploration (EDA):**

   After collecting the data, we study it to understand its structure and values. We use charts, graphs, and statistical methods to find patterns, trends, and unusual values.

4. **Data Cleaning and Preprocessing:**

   In this step, we clean the data by handling missing values, removing duplicate records, and correcting errors. We also prepare the data in a suitable format for analysis and model building.

5. **Feature Selection and Engineering:**

   Feature selection means choosing the important columns needed for the project. Feature engineering means creating or changing columns to make the data more useful for building a Machine Learning model.

6. **Model Building and Evaluation:**

   In this step, we choose a suitable Machine Learning algorithm and train the model using the prepared data. We then check its performance using measures such as Accuracy, Precision, Recall, and F1-Score.

7. **Deployment and Maintenance:**

   After testing the model, we make it available for real use through an application, API, or dashboard. We also check its performance regularly and update it when needed.



## Question 3: Frame a Machine Learning Problem Statement

### Title: Customer Churn Prediction for a Consumer Company

### Objective

The main objective of this project is to build a Machine Learning model that can predict whether a customer will continue using a company's services or stop using them. The model uses customer details, purchase history, and activity records to make predictions.

### Problem Type

This is a Supervised Binary Classification problem.

The target variable is `Churn`.

- `1` means the customer has left the company.
- `0` means the customer is still using the company's services.

### Input Features

The model uses different types of customer information, such as:

- **Customer Details:** Age, gender, and location.
- **Financial Details:** Annual income and account balance.
- **Purchase Details:** Total purchases, average order value, purchase frequency, and time since the last purchase.
- **Customer Activity:** Customer activity and support service records.

### Business Value

Customer Churn Prediction helps a company find customers who may stop using its services. The company can contact these customers and offer suitable discounts, better services, or loyalty programs.

This helps the company keep more customers, reduce customer loss, improve customer satisfaction, and increase its profit over time.



## Question 4: Explain Tensors and Give Examples Using NumPy

### What is a Tensor?

A tensor is a data structure used to store numbers in one or more dimensions. It is a general form of a scalar, vector, and matrix.

Tensors are widely used in Machine Learning and Deep Learning. They help store input data, model values, and calculation results. Libraries such as TensorFlow and PyTorch use tensors to perform different calculations.

### Types of Tensors Based on Dimensions

1. **Scalar (0-D Tensor):**

   A scalar is a single numerical value. It has zero dimensions.

   Example: `45` or `4.5`

2. **Vector (1-D Tensor):**

   A vector is a list of numerical values. It has one dimension.

   Example: `[25, 30, 45]`, which can represent the ages of customers.

3. **Matrix (2-D Tensor):**

   A matrix is a collection of values arranged in rows and columns. It has two dimensions.

   Example: A table containing customer ages, incomes, and purchase details.

4. **3-D and Higher-Dimensional Tensors:**

   These tensors store data in three or more dimensions. They are useful for working with images, videos, and time-based data.

   Example: A colour image can be represented using height, width, and colour channels.

### Python Code Examples Using NumPy

NumPy is a Python library used to work with arrays and perform mathematical calculations. We can use NumPy to create different types of tensors and check their dimensions.

### 1. Scalar (0-D Tensor)

A scalar stores a single numerical value.

```python
import numpy as np

scalar_tensor = np.array(45)

print("0-D Tensor (Scalar):", scalar_tensor)
print("Dimensions:", scalar_tensor.ndim)
```

### 2. Vector (1-D Tensor)

A vector stores a list of numerical values.

```python
vector_tensor = np.array([22, 35, 40, 28, 52])

print("1-D Tensor (Vector):", vector_tensor)
print("Dimensions:", vector_tensor.ndim)
```

### 3. Matrix (2-D Tensor)

A matrix stores values in rows and columns.

```python
matrix_tensor = np.array([
    [25, 50000, 5],
    [32, 72000, 8],
    [47, 120000, 12]
])

print("2-D Tensor (Matrix):\n", matrix_tensor)
print("Dimensions:", matrix_tensor.ndim)
print("Shape:", matrix_tensor.shape)
```

### 4. 3-D Tensor

A 3-D tensor stores data in three dimensions.

```python
tensor_3d = np.array([
    [[25, 50000], [32, 72000]],
    [[47, 120000], [19, 30000]]
])

print("3-D Tensor:\n", tensor_3d)
print("Dimensions:", tensor_3d.ndim)
print("Shape:", tensor_3d.shape)
```

### Explanation of the Code

- `import numpy as np` is used to import the NumPy library.
- `np.array()` is used to create an array.
- `ndim` tells us the number of dimensions of an array.
- `shape` tells us the size of the array in each dimension.
- A scalar has zero dimensions, a vector has one dimension, a matrix has two dimensions, and the last example has three dimensions.

### Conclusion

Tensors are important in Machine Learning and Deep Learning because they help store and process numerical data. NumPy makes it easy to create arrays and understand their dimensions and shapes.


## Part B: Data Acquisition

````python
# 5.Import data from different sources
import pandas as pd
import numpy as np

# 1. Real Customer Churn Dataset Generation with real-world characteristics
np.random.seed(42)
n_samples = 500

customer_ids = [f"CUST-{i:04d}" for i in range(1, n_samples + 1)]
ages = np.random.randint(18, 70, size=n_samples)
genders = np.random.choice(['Male', 'Female'], size=n_samples, p=[0.52, 0.48])
incomes = np.random.randint(20000, 180000, size=n_samples)
purchases = np.random.randint(1, 45, size=n_samples)
tenure_months = np.random.randint(1, 72, size=n_samples)
churns = np.random.choice([0, 1], size=n_samples, p=[0.75, 0.25])

# Initialize Dataset dictionary 
df_real = pd.DataFrame({
    'CustomerID': customer_ids,
    'Age': ages,
    'Gender': genders,
    'AnnualIncome': incomes,
    'TotalPurchases': purchases,
    'TenureMonths': tenure_months,
    'Churn': churns
})

# Add some missing values to simulate real-world data
df_real.loc[np.random.choice(df_real.index, 15), 'AnnualIncome'] = np.nan
df_real.loc[np.random.choice(df_real.index, 10), 'Age'] = np.nan

# Add some duplicate rows to simulate real-world data
df_real = pd.concat([df_real, df_real.iloc[0:5]], ignore_index=True)

# Save as CSV file
df_real.to_csv('real_customer_churn_data.csv', index=False)
print("✅ Real Project CSV File 'real_customer_churn_data.csv' Created Successfully!")

# 2. Load CSV file using Pandas
df_csv = pd.read_csv('real_customer_churn_data.csv')
print("\n--- First 5 Rows of Real Dataset ---")
print(df_csv.head())
````

````python
import json
import sqlite3
import requests

# 2. Parse a JSON file
# Creating a JSON file from our raw data and then parsing it
df_real.head(15).to_json('customer_data.json', orient='records', indent=4)
print("✅ JSON File 'customer_data.json' created successfully!")

# Load/Parse JSON file using Pandas
df_json = pd.read_json('customer_data.json')
print("\n--- Parsed JSON Data Head ---")
display(df_json.head(3))
````

````python
# 3. Connect to a SQL table and fetch records
# Creating an SQLite database and storing the customer records in a SQL table
conn = sqlite3.connect('company_database.db')
df_csv.to_sql('sql_customer_table', conn, index=False, if_exists='replace')

# Querying the SQL table using Pandas
query = "SELECT CustomerID, Age, AnnualIncome, Churn FROM sql_customer_table WHERE Age > 30 LIMIT 5"
df_sql = pd.read_sql(query, conn)
print("\n--- Fetched Records from SQL Database (Age > 30) ---")
display(df_sql)
conn.close()
````

````python
# 4. Fetch data from an API - Using JSONPlaceholder Users API as a dummy REST API
try:
    api_url = 'https://jsonplaceholder.typicode.com/users'
    response = requests.get(api_url)
    
    if response.status_code == 200:
        api_data = response.json()
        # Flattening JSON response into a Pandas DataFrame
        df_api = pd.json_normalize(api_data)
        print("\n--- Fetched Data from Public API ---")
        display(df_api[['id', 'name', 'email', 'phone']].head(3))
    else:
        print("Failed to fetch API data. Status code:", response.status_code)
except Exception as e:
    print("An error occurred during API fetching:", e)
````

## Part C: Data Understanding & Cleaning

````python
# 6. Initial Exploration
print("--- Cleaned Head (First 5 Rows) ---")
display(df_csv.head())

print("\n--- Dataset Info ---")
df_csv.info()

print("\n--- Descriptive Statistics ---")
display(df_csv.describe())

print("\n--- Missing Values Count Per Column ---")
print(df_csv.isnull().sum())

print("\n--- Total Duplicate Rows ---")
print(f"Duplicates: {df_csv.duplicated().sum()}")
````

````python
# 7. Apply Data Cleaning

# Remove duplicates
initial_count = len(df_csv)
df_cleaned = df_csv.drop_duplicates().copy()
print(f"✅ Removed {initial_count - len(df_cleaned)} duplicate rows.")

# Handle missing values via Imputation
# Impute AnnualIncome with median
income_median = df_cleaned['AnnualIncome'].median()
df_cleaned['AnnualIncome'] = df_cleaned['AnnualIncome'].fillna(income_median)

# Impute Age with mean
age_mean = df_cleaned['Age'].mean()
df_cleaned['Age'] = df_cleaned['Age'].fillna(age_mean)

# Verification after cleaning
print("\n--- Missing Values After Cleaning ---")
print(df_cleaned.isnull().sum())

print("\n--- Cleaned Dataset Preview ---")
display(df_cleaned.head())
````

## Part D: Exploratory Data Analysis (EDA)

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Set style for plots
sns.set_theme(style="whitegrid")
plt.figure(figsize=(15, 12))

# 8. Univariate Analysis (Distribution Plots) 
# Distribution of Age
plt.subplot(3, 2, 1)
sns.histplot(df_cleaned['Age'], kde=True, color='blue', bins=20)
plt.title('Univariate: Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')

# Distribution of Annual Income
plt.subplot(3, 2, 2)
sns.histplot(df_cleaned['AnnualIncome'], kde=True, color='green', bins=20)
plt.title('Univariate: Annual Income Distribution')
plt.xlabel('Annual Income')
plt.ylabel('Frequency')
````

````python
# --- 2. Bivariate Analysis ---
# Relationship between Gender & Purchases
plt.subplot(3, 2, 3)
sns.boxplot(x='Gender', y='TotalPurchases', data=df_cleaned, hue='Gender', palette='Set1', legend=False)
plt.title('Bivariate: Gender vs Total Purchases')
plt.xlabel('Gender')
plt.ylabel('Total Purchases')

# Relationship between Income & Churn
plt.subplot(3, 2, 4)
sns.boxplot(x='Churn', y='AnnualIncome', data=df_cleaned, hue='Churn', palette='Pastel1', legend=False)
plt.title('Bivariate: Annual Income vs Churn')
plt.xlabel('Churn (0 = Retained, 1 = Churned)')
plt.ylabel('Annual Income')
````

````python
# 10. Multivariate Analysis (Correlation Heatmap) 
plt.subplot(3, 2, 5)
numerical_df = df_cleaned.select_dtypes(include=['int64', 'float64'])
corr_matrix = numerical_df.corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Multivariate: Correlation Heatmap')

plt.tight_layout()
plt.show()

# --- Pairplot for feature interactions ---
print("\n--- Generating Pair Plot for Feature Interactions ---")
sns.pairplot(df_cleaned[['Age', 'AnnualIncome', 'TotalPurchases', 'TenureMonths', 'Churn']], hue='Churn', palette='husl')
plt.show()
````
